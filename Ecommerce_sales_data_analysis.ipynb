```python
# Preserve data-quality flags
df_clean['age_was_invalid'] = (
    df_clean['age'].isna() |
    (df_clean['age'] < 1) |
    (df_clean['age'] > 100)
)

df_clean['quantity_was_invalid'] = (
    df_clean['quantity'].isna() |
    (df_clean['quantity'] < 1) |
    (df_clean['quantity'] > 10)
)

df_clean['shipping_was_invalid'] = (
    df_clean['days_to_ship'].isna() |
    (df_clean['days_to_ship'] < 0)
)

df_clean['shipping_delay_flag'] = df_clean['days_to_ship'] > 30

# Calculate medians using valid observations
age_median = df_clean.loc[
    df_clean['age'].between(1, 100), 'age'
].median()

quantity_median = df_clean.loc[
    df_clean['quantity'].between(1, 10), 'quantity'
].median()

shipping_median = df_clean.loc[
    df_clean['days_to_ship'].between(0, 30), 'days_to_ship'
].median()

# Replace missing and invalid values
df_clean.loc[
    ~df_clean['age'].between(1, 100), 'age'
] = age_median

df_clean.loc[
    ~df_clean['quantity'].between(1, 10), 'quantity'
] = quantity_median

df_clean['discount_pct'] = df_clean['discount_pct'].fillna(
    df_clean['discount_pct'].median()
)

df_clean['customer_satisfaction'] = df_clean['customer_satisfaction'].fillna(
    df_clean['customer_satisfaction'].median()
)

df_clean.loc[
    df_clean['days_to_ship'].isna() |
    (df_clean['days_to_ship'] < 0), 'days_to_ship'
] = shipping_median

print("Numerical cleaning completed!")
print("Remaining missing values:")
print(df_clean.isna().sum())
```

```python
# Remove unwanted spaces from all text columns
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].str.strip()

# Standardize gender
df_clean['gender'] = (
    df_clean['gender']
    .str.lower()
    .replace({'m': 'male', 'f': 'female'})
    .str.title()
)

# Standardize order status
df_clean['order_status'] = df_clean['order_status'].str.title()

print("Gender categories:")
print(df_clean['gender'].value_counts())

print("\nOrder status categories:")
print(df_clean['order_status'].value_counts())
```


## Conclusion

This project analyzed 4,200 e-commerce transactions using Python, Pandas, NumPy, Matplotlib and Seaborn.

The analysis explored sales performance, profitability, product categories, regional performance, customer satisfaction, discounts and order returns.

The key findings were:

- Total recorded sales were approximately 264.27 million monetary units.
- Total recorded profit was approximately 49.52 million monetary units.
- Electronics generated the highest sales among product categories.
- Furniture achieved a profit margin of approximately 28.22%.
- The South region recorded the highest sales.
- 2023 was the strongest year in terms of sales.
- Tablets were the highest-selling product by recorded sales.
- Average customer satisfaction was 3.02 out of 5.
- The overall order return rate was 14.57%.

## Business Recommendations

1. **Improve customer satisfaction:** Investigate customer feedback and identify the reasons behind moderate satisfaction scores.
2. **Analyze product returns:** Examine return patterns by product, category and region to identify potential causes.
3. **Optimize product strategy:** Study the strong sales performance of Electronics and the profitability of Furniture.
